```python
# set better to use it with math
a={1,2,3,4,5,6}
b={1,4,5,8,9,0}

#union 
print(a|b)
print(a.union(b))
print(b.union(a))
```

```python
#intersection
print(b.intersection(a))
print(a&b)
print(a.intersection(b))
```

```python
# differnce of two sets
print(a-b)
print(a.difference(b))
print()
print(b-a)
print(b.difference(a))
```

```python
# symmetric differnce 
print(a^b)
print(a.symmetric_difference(b))
```
