**To filter the rows based on such a function, use the conditional function inside the selection brackets []. In this case, the condition inside the selection brackets df["Country"].isin(['USA', 'Canada']) checks for which rows the Country column is either 'USA' or 'Canada'**

```python
class_USA_Cand = df[df["Country"].isin(['USA','Canada'])]
class_USA_Cand
```

```python
class_USA_Cand = df[(df['Country']=='USA')|(df['Country']=='Canada')]
class_USA_Cand
```

*When combining multiple conditional statements, each condition must be surrounded by parentheses (). Moreover, you can not use or/and but need to use the or operator | and the and operator &.*

**The notna() conditional function returns a True for each row the values are not a Null value. As such, this can be combined with the selection brackets [] to filter the data table.**

```python
not_null = df[df['OrderDate'].notna()]
not_null.head()
```

***How do I select specific rows and columns from a DataFrame?***

#### **loc vs iloc in Pandas — In Depth** 

- loc → selects using labels/names
- iloc → selects using integer positions

```python
print(df.loc[1,'Price']) #[row,column]
```

**Give me the value whose label is 1 and whose column label is "price"**

```python
# if you know the order or the index of the column you can do that 
print(df.iloc[1,5])
```

```python
ndf = pd.DataFrame({
    "name": ["Ali", "Ahmed", "Omar", "Mona"],
    "age": [20, 22, 21, 23],
    "city": ["Cairo", "Giza", "Luxor", "Aswan"]
})

ndf
```

```python
# if i wanna select spacific row 
print(ndf.loc[1])
```

```python
ndf.loc[[1,2]] # select rows  1 and  2 
```

**Selecting columns with loc**


```python
ndf.loc[:,'age'] # the " : " -> means all rows
```

**selecting multiple columns with loc**

```python
ndf.loc[:,['name','age']]
```

**selecting multiple rows and multiple columns**

```python
ndf.loc[[1,3],['name','age']]
```

```python
ndf.loc[1:3,['name','age']] # you have to notice that 3 is included
```

```python
ndf.loc[:, "name":"city"]
```

```python
ndf[ndf['age']>21]
```

```python
ndf.loc[ndf['age']>21]
```

```python
ndf.loc[ndf["age"] > 21, ["name", "age"]] # Condition + specific columns
```

```python
## multiple conditions
ndf.loc[
    (ndf["age"] > 20) & (ndf["city"] == "Giza")
]
```

```python
ndf.loc[
    (ndf["city"] == "Giza") | (ndf["city"] == "Cairo")
]
```

##### **loc** can modify data

```python
ndf.loc[ndf["age"] > 21, "city"] = "Unknown"
ndf
```

```python
#Changing multiple columns
ndf.loc[ndf['age']>21,['age','city']]=[55,'A/n']
ndf
```

### iloc() ==> means : integer location

**df.iloc[row_position, column_position]**

```python
## the key while using iloc() ==> just use index

ndf.iloc[1] ## accessing row with index 1
```

```python
## accessing mutilpe rows
ndf.iloc[[1,2]]
```

```python
# accessing range of rows
ndf.iloc[1:3]  # you have to notice something the with index 3 is not include as iloc deal with indexes not lables
```

```python
## accessing row and columns
ndf.iloc[:,2]
```

```python
## multiple columns 
ndf.iloc[:,[1,2]]
```

```python
## Rows + columns
ndf.iloc[[0, 2], [0, 1]]
```

|                   | `loc`           | `iloc`       |
| ----------------- | --------------- | ------------ |
| Uses              | Labels          | Positions    |
| `df.loc[1]`       | label 1         | ❌            |
| `df.iloc[1]`      | ❌               | position 1   |
| Slice ending      | **Included**    | **Excluded** |
| Conditions        | ✅               | ❌ generally  |
| Column names      | ✅               | ❌            |
| Integer positions | Not its purpose | ✅            |

### How to calculate summary statistics

```python
df['Price'].mean()
```

```python
df['Price'].describe()
```

**agg() function allow you to select the columns and statatical method you want**

```python
df.agg(
    {
        "Price": ["min", "max", "median", "skew"],
        "Quantity": ["min", "max", "median", "mean"],
    }
)
```

**Aggregating statistics grouped by category**

```python
df[['Price','Category']].groupby('Category').mean()
```

```python
df.groupby("Category").mean(numeric_only=True)
```

```python
# display the mean of the total_price group by Category & country 
df.groupby(['Category','Country'])['total_price'].mean()
```

**Count number of records by category**

*The value_counts() method counts the number of records for each category in a column.*

```python
df['Category'].value_counts()
```

```python
# you can do it using groupby
df.groupby("Category")["Category"].count()
```

### VIP note:
**Both size and count can be used in combination with groupby. Whereas size includes NaN values and just provides the number of rows (size of the table), count excludes the missing values. In the value_counts method, use the dropna argument to include or exclude the NaN values.**


REMEMBER
Aggregation statistics can be calculated on entire columns or rows.

groupby provides the power of the split-apply-combine pattern.

value_counts is a convenient shortcut to count the number of entries in each category of a variable.
