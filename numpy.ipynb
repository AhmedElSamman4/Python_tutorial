### how to change data type of array

```python
myarray1=np.array([1,3,4,5,6])
myarray2=np.array([1.5,4.5,6.7])
myarray3=np.array(['ahmed','b','khaled'])

print(myarray1.dtype)
print(myarray2.dtype)
print(myarray3.dtype) # U6 ==> Unicode string , and the max lenght is of string is 6
```

```python
# how to change data type of an array
myarray1=myarray1.astype(float)
print(myarray1.dtype)
```

```python
myarray4=np.array([3,4,5,6],dtype=float)# float or f
print(myarray4)
print(myarray4.dtype)
```

```python
myarray5=np.array([2.3,4.5,6.7],dtype=int) # int or i
print(myarray5)
print(myarray5.dtype)
```

```python
## changing to bool type
myarrayb = np.array([0,4,0,6,1])
print(myarrayb)
```
