```python
name = 'ahmed'
age = 44
```

```python
print('my name is',name , 'my age is',age)
```

```python
# {} is a replacement fields ==> placeholder
'my name is {} and my age is {}'.format(name,age)
```

```python
# here we pass the index of the value in the format to be more spacified 
'my name is {0} and my age is {1}'.format(name,age) # format (0,1,2,....etc)
```

```python
'the value of by is {0:.4f}'.format(3.1415926) ## .4f means that print 4 digits after fraction
```

```python
'my age is {0:03}'.format(22) ## :03 means that i wanna 3 zero's 
```

```python
'my grandfather age is {0:03}'.format(102)
```

### Formating F-string

```python
name = 'mohamed'
age = 45
print(f'name is {name} and age is {age}')
```

```python
print(f'{23.934898:.2f}') # f-> float
```

```python
print(f'{12391:07d}') # 7 zero's , d -> int
```

```python
print(f'{65:c}') # char 
```

```python
from decimal import Decimal
```

```python
print(f'{Decimal(1000000000000000000.0):.2e}') # 18 = num of zero's
```

```python
# grouping digits 
f'{Decimal(1000000000000000000):,}'
```

```python
f'{28738273:,d}'
```

#### splitting and joining string

```python
newtext = 'do it again and again'
newtext = newtext.split(' ')
newtext
```

```python
ctext = 'do, it, again, and, again'
ctext = [x.strip() for x in ctext.split(',')]
ctext

```

```python
# line splitting 
lines= '''you are my favourite broo
see you soon 
take care'''
lines 
```

```python
lines.splitlines()
```

```python
# str join text 
xlist = ['coffee','you','and', 'me']
" ".join(xlist)
```

```python
# partition divid your string into to segment at the first delimeter you spacified
tt = 'you VS me; on friday at 7 am'.partition(';')
tt
## thus return tuple
```
